```python
import heapq

def greedy_best_fit_search(graph, heuristic, start, goal):
    pq = []
    heapq.heappush(pq, (heuristic, start, [start]))

    visited = set()
    while pq:
        h_val, current_node, path = heapq.heappop(pq)

        if current_node == goal:
            return path

        if current_node not in visited:
            visited.add(current_node)
            for neighbor, weight in graph.get(current_node , []):
                if neighbor not in visited:
                    heapq.heappush(pq, (heuristic[neighbor], neighbor, path + [neighbor]))
    return None

graph = {
    'A' : [('B', 3), ('C', 2)],
    'B' : [('D', 4), ('E', 6)],
    'C' : [('F', 1)],
    'D' : [],
    'E' : [('G', 2)],
    'F' : [('G', 5)],
    'G' : []
}

heuristic = {
    'A':10,
    'B':8,
    'C':5,
    'D':7,
    'E':3,
    'F':3,
    'G':0
}

start_node = 'A'
goal_node = 'G'

path = greedy_best_fit_search(graph, heuristic, start_node, goal_node)

if(path):
    print("Path Found: ", (' -> ').join(path))
else:
    print("No path found")
```
