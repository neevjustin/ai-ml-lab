```python
import heapq
```

```python
def uniform_cost_search(graph, start, goal):
    visited_costs = {start: 0}

    priority_queue = [(0, start, [start])]

    while priority_queue:
        current_cost, current_node, path = heapq.heappop(priority_queue)
        print(f"Popped node : '{current_node}' with cost {current_cost} and path {path}")

        if current_node == goal:
            return current_cost, path

        if current_cost > visited_costs.get(current_node, float('inf')):
            continue

        for edge_cost, neighbor in graph.get(current_node, []):
            new_cost = current_cost + edge_cost

            if neighbor not in visited_costs or new_cost < visited_costs[neighbor]:
                visited_costs[neighbor] = new_cost
                updated_path = path + [neighbor]

                heapq.heappush(priority_queue, (new_cost, neighbor, updated_path))

    return float("inf"), []
```

```python
graph = {
        'A':[(1, 'B'), (4, 'C')],
        'B':[(2, 'C'), (5, 'D'), (12, 'E')],
        'C':[(2, 'D')],
        'D':[(3, 'E')],
        'E':[]
    }
cost, path = uniform_cost_search(graph, 'A', 'E')

if cost == float("inf"):
    print("Target unreachable from the start node")
else:
    print(f"Optimal Path : {(' -> ').join(path)} with cost {cost}")
                 
```
